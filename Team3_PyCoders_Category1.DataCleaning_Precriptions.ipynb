```python
import pandas as pd
import numpy as np

df = pd.read_csv("/Users/are/Development/DA229/PythonHackathon/Python_Hackathon_Sep_2026/cardiac_failure/patient_precriptions.csv")

# 1) Basic inspection
print(df.shape)
print(df.head())
print(df.isna().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate patient-drug pairs:", df.duplicated(subset=["inpatient_number", "drug_name"]).sum())
```

```python
# 2) Clean patient ID
df = df.rename(columns={'inpatient_number': 'patient_id'})
```

```python
# 3) Clean drug names
# Keep as string so text cleaning and grouping does not coerce it to numeric
# or create unexpected dtype changes.
df["drug_name"] = df["drug_name"].astype("string")

df["drug_name"] = df["drug_name"].str.strip()df["drug_name"] = df["drug_name"].str.lower()
df["drug_name"] = df["drug_name"].str.replace(r"\s+", " ", regex=True)
```

```python
# 4) Remove blank or missing drug entries
df = df[df["drug_name"].notna() & df["drug_name"].str.strip().ne("")]
```
